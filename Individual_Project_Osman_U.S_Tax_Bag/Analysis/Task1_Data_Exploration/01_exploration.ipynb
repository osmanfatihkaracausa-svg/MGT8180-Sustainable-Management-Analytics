```python
# Import the necessary libraries
import pandas as pd
import numpy as np

# Confirm that the libraries are imported successfully
print("pandas version:", pd.__version__)
print("numpy version:", np.__version__)
```

```python
# Load the Bag Tax Dataset from the CSV file
df = pd.read_csv('../../Data/Bag_Tax_Data.csv', encoding='utf-8-sig')
# Display the first few rows of the dataset to confirm it loaded correctly
df.head(10) 
```

```python
# Inspect the structure of the dataset

# 1. Shape - number of rows and columns
print("Dataset shape(rows, columns):", df.shape)
print()
# 2. Columns - list of columns names
print("Column Names:")
print(list(df.columns))
print()
# 3. Data Types - Data types of each columns
print("Data Types:")
print(df.dtypes)
print()
# 4. Missing Values - Checking for missin values in each columns
print("Missing values per column:")
print(df.isna().sum())
```

```python
# Show full missing value counts for each columns (not truncated)
pd.set_option('display.max_rows', None) #show all rows in the output
print("Missing values per column :")
print(df.isna().sum())
pd.reset_option('display.max_rows') #reset to default (10) rows
```

### Missing data observations 

- The dataset contains **71,973 records** and **14 columns**.
- Only one column has missing values: **City** (22,956 missing values, ~32% of records). But they have locations.
- All other columns are complete (0 missing values).

### Decision

I will **keep all rows** and **not impute** the missing City values. Reason:
- City is not required for any analysis task (Bag Count, Amount Collected, vendor analysis, time trends).
- The **State** and **Zip code** columns are fully populated, providing complete geographic information.
- All transactions in this dataset occurred at retail establishments, regardless of vendor billing address.

```python
# Since there are no missing values, I realized Date format is not in datetime format. I will convert it to datetime format for easier analysis later on.
# Step 1: Convert 'Date' Column to Datetime Format
df['date From'] = pd.to_datetime(df['Date From'], errors='coerce') # Convert to datetime, coerce errors to NaT
df['Date To'] = pd.to_datetime(df['Date To'], errors='coerce') 
# Step 2: What is the date range of the dataset?
print("=" * 60)
print("DATE RANGE CHECK")
print("=" * 60)
print(f"Date From — earliest: {df['Date From'].min()}")
print(f"Date From — latest:   {df['Date From'].max()}")
print(f"Date To   — earliest: {df['Date To'].min()}")
print(f"Date To   — latest:   {df['Date To'].max()}")
```

```python
# Re-convert all date columns from text to datetime
df['Date From'] = pd.to_datetime(df['Date From'], errors='coerce')
df['Date To'] = pd.to_datetime(df['Date To'], errors='coerce')
df['Create Date'] = pd.to_datetime(df['Create Date'], errors='coerce')

# Verify the conversion worked — check the data type
print("Data types after conversion:")
print(df[['Date From', 'Date To', 'Create Date']].dtypes)
print()
print("=" * 60)
print("DATE RANGE — VERIFIED")
print("=" * 60)
print(f"Date From — earliest: {df['Date From'].min()}")
print(f"Date From — latest:   {df['Date From'].max()}")
print(f"Date To   — earliest: {df['Date To'].min()}")
print(f"Date To   — latest:   {df['Date To'].max()}")
```

```python
# Show the actual bad rows
print("=" * 60)
print("BAD ROWS — Date From before 2012-01-01")
print("=" * 60)
bad_old = df[df['Date From'] < '2012-01-01']
print(f"Total bad rows: {len(bad_old)}")
print()
print(bad_old[['File ID', 'Date From', 'Date To', 'Vendor Name', 'Bag Count']].head(10))
print()
print("=" * 60)
print("BAD ROWS — Date To after 2025-12-31")
print("=" * 60)
bad_future = df[df['Date To'] > '2025-12-31']
print(f"Total bad rows: {len(bad_future)}")
print()
print(bad_future[['File ID', 'Date From', 'Date To', 'Vendor Name', 'Bag Count']].head(10))
```

```python
# ============================================================
# Cleaning Step: Remove rows with invalid dates
# ============================================================
# Justification:
# - Montgomery County's open data portal states the dataset 
#   contains records "since this law went into effect" (Jan 1, 2012)
# - Rows with Date From before 2012-01-01 are outside this scope
# - Rows with Date To far in the future (2027+) are data entry errors

# Save the original count for comparison
original_count = len(df)
print(f"Original record count: {original_count:,}")
print()

# Define what counts as a valid record
valid_mask = (
    (df['Date From'] >= '2012-01-01') &     # Tax started here
    (df['Date To'] <= '2025-12-31') &       # Reasonable upper bound
    (df['Date From'] <= df['Date To'])      # From must be before To
)

# Apply the filter — keep only valid rows
df_clean = df[valid_mask].copy()

# Report what was removed
removed = original_count - len(df_clean)
print(f"Rows removed:           {removed:,}")
print(f"Records after cleaning: {len(df_clean):,}")
print(f"Percentage retained:    {len(df_clean)/original_count*100:.2f}%")
```

```python
# ============================================================
# Verification: Confirm bad rows are gone
# ============================================================

print("BEFORE cleaning:")
print(f"  Date From — earliest: {df['Date From'].min()}")
print(f"  Date From — latest:   {df['Date From'].max()}")
print(f"  Date To   — earliest: {df['Date To'].min()}")
print(f"  Date To   — latest:   {df['Date To'].max()}")
print()
print("AFTER cleaning:")
print(f"  Date From — earliest: {df_clean['Date From'].min()}")
print(f"  Date From — latest:   {df_clean['Date From'].max()}")
print(f"  Date To   — earliest: {df_clean['Date To'].min()}")
print(f"  Date To   — latest:   {df_clean['Date To'].max()}")
```

```python
# ============================================================
# Save the cleaned dataset for use in Tasks 2, 3, 4
# ============================================================
import os

# Create the clean folder if it doesn't exist
clean_folder = '../../Data/clean'
os.makedirs(clean_folder, exist_ok=True)

# Save the cleaned data
output_path = f'{clean_folder}/Bag_Tax_Clean.csv'
df_clean.to_csv(output_path, index=False)

print(f"✅ Cleaned data saved to: {output_path}")
print(f"   Rows: {len(df_clean):,}")
print(f"   Columns: {len(df_clean.columns)}")
```
